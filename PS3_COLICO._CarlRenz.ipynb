```python
#!pip install pgmpy

from pgmpy.models import BayesianNetwork
from pgmpy.factors.discrete.CPD import TabularCPD
from pgmpy.inference import VariableElimination
```

# Setting up your model

### First, set the structure

```python

model = BayesianNetwork([('IQ Level', 'Marks'),
                         ('IQ Level', 'Apti Score'),
                         ('Exam Level', 'Marks'),
                         ('Marks', 'Admission')])
```

### Then set up the relationships (the CPDs) and the probability table

```python
IQ_Level = TabularCPD(
                variable = 'IQ Level',
                variable_card = 2,
                values = [[0.8], [0.2]])

Apti_Score = TabularCPD(
                variable = 'Apti Score',
                variable_card = 2,
                values = [[0.75, 0.4],
                          [0.25, 0.6]],
                evidence = ['IQ Level'],
                evidence_card = [2])

Exam_Level = TabularCPD(
                variable = 'Exam Level',
                variable_card = 2,
                values = [[0.7], [0.3]])

                          
Marks = TabularCPD(
                variable = 'Marks',
                variable_card = 2,
                values = [[0.6, 0.9, 0.5, 0.8],
                          [0.4, 0.1, 0.5, 0.2]],
                evidence = ['Exam Level', 'IQ Level'],
                evidence_card = [2, 2])

Admission = TabularCPD(
                variable = 'Admission',
                variable_card = 2,
                values = [[0.6, 0.9],
                          [0.4, 0.1]],
                evidence = ['Marks'],
                evidence_card = [2])
```

IQ_Level has 2 var.card and Exam Level has also a 2 var.card so this make Marks to have a (2,2) eveidence card

While Apti and Admission has only 1 parent which is IQ level and Marks respectively as mention. Thats why it turned out to have a [2] as both of Apti and Admission evidence card.

On the otherhand IQ Level and Exam level having no parent thus making them to have a non-evidence card and they have a 2 variable card.

### Add the relationships to your models

```python
model.add_cpds(IQ_Level, Apti_Score, Exam_Level, Marks, Admission)
```

here we can see our model added is used in a context method of .add cpds this will define the probability of the given node with its parent we inputted.

adding the model variable for each cpds var. added to the model.

# Making inferences

### We can get probability distributions that are not explicitly spelled out in our graphs

This mean that we make a inference between variables that is really not connected to each other or node doesnt seem to appear in the given Bayesian Network Graph or does not really give an inference to each of the input variables.
This will let us exploread and changed and combined variables that our not explicititly spelled or shown in our bayesian network graphs.

```python
model_infer = VariableElimination(model)
```

```python
prob_appointment = model_infer.query(variables = ['Admission', 'Marks'])
print(prob_appointment)
```

```python
prob_appointment = model_infer.query(variables = ['Apti Score', 'Marks'])
print(prob_appointment)
```

```python
prob_appointment = model_infer.query(variables = ['Admission', 'Apti Score', 'Marks'])
print(prob_appointment)
```

```python
prob_appointment = model_infer.query(variables = ['Exam Level', 'Admission'])
print(prob_appointment)
```

```python
prob_appointment = model_infer.query(variables = ['Exam Level', 'IQ Level'])
print(prob_appointment)
```

```python
prob_appointment = model_infer.query(variables = ['Admission', 'IQ Level'])
print(prob_appointment)
```

```python
prob_appointment = model_infer.query(variables = ['Exam Level','IQ Level', 'Admission'])
print(prob_appointment)
```

### We can also get conditional probability distributions that take into account what we already know

```python
prob = model_infer.query(variables = ['Admission'], 
                         evidence = {'IQ Level':1, 'Apti Score':1})
print(prob)
```

Here what this code structure do is it is asking for a query where in it takes into account based on the given evidence we input so here the distribution of variable of probability we are finding is the  Admission within the based of evidence of  node IQ level and Exam level

```python
prob = model_infer.query(variables = ['Admission'], 
                         evidence = {'Marks':0})
print(prob)
```

thus findings analyze m0 = 60% and m1 = 40%.
