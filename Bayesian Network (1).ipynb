```python
#!pip install pgmpy

from pgmpy.models import BayesianNetwork
from pgmpy.factors.discrete.CPD import TabularCPD
from pgmpy.inference import VariableElimination
```

# Setting up your model

### First, set the structure

```python
model = BayesianNetwork([('Exam level', 'Marks'),
                         ('IQ level', 'Marks'),
                         ('IQ level', 'Apti. score'),
                         ('Marks', 'Admission')])
```

### Then set up the relationships (the CPDs) and the probability table

```python
Exam_level = TabularCPD(
                variable = 'Exam level',
                variable_card = 2,
                values = [[0.7], [0.3]])

IQ_level = TabularCPD(
                variable = 'IQ level',
                variable_card = 2,
                values = [[0.8], [0.2,]])

Marks = TabularCPD(
                variable = 'Marks',
                variable_card = 2,
                values = [[0.6, 0.9, 0.5, 0.8,] ,
                          [0.4, 0.1, 0.5, 0.2]],
                evidence = ['Exam level', 'IQ level'],
                evidence_card = [2, 2])

Apti_score = TabularCPD(
                variable = 'Apti. score',
                variable_card = 2,
                values = [[0.75, 0.4],
                          [0.25, 0.6]],
                evidence = ['IQ level'],
                evidence_card = [2])
Admission = TabularCPD(
                variable = 'Admission',
                variable_card = 2,
                values = [[0.6, 0.9],
                          [0.4, 0.1]],
                evidence = ['Marks'],
                evidence_card = [2])
```

### Add the relationships to your models

```python
model.add_cpds(Exam_level, IQ_level, Marks, Apti_score, Admission)
```

### Examine the structure of your graph

```python
model.get_cpds()
```

### Find local independencies

```python
model.get_independencies()
```

### Find active trail nodes

```python
model.active_trail_nodes('Exam level')
model.active_trail_nodes('IQ level')
model.active_trail_nodes('Marks')
model.active_trail_nodes('Apti. score')
model.active_trail_nodes('Admission')
```

```python
model.local_independencies('Exam level')
model.local_independencies('IQ level')
model.local_independencies('Marks')
model.local_independencies('Apti. score')
model.local_independencies('Admission')
```

### Get all independencies

```python
model.get_independencies()
```

# Making inferences

### We can get probability distributions that are not explicitly spelled out in our graphs

```python
model_infer = VariableElimination(model)
```

```python
prob_appointment = model_infer.query(variables = ['Admission', 'Marks'])
print(prob_appointment)
```

### We can also get conditional probability distributions that take into account what we already know

```python
prob = model_infer.query(variables = ['Admission'], 
                         evidence = {'Marks':1})
print(prob)
```

```python
prob = model_infer.query(variables = ['Admission'], 
                         evidence = {'Marks':0})
print(prob)
```

```python
prob = model_infer.query(variables = ['Marks'], 
                         evidence = {'IQ level':0, 'Exam level': 0})
print(prob)
```

```python
prob = model_infer.query(variables = ['Marks'], 
                         evidence = {'IQ level':1, 'Exam level': 0})
print(prob)
```

```python
prob = model_infer.query(variables = ['Marks'], 
                         evidence = {'IQ level':0, 'Exam level': 1})
print(prob)
```

```python
prob = model_infer.query(variables = ['Marks'], 
                         evidence = {'IQ level':1, 'Exam level': 1})
print(prob)
```

```python
prob = model_infer.query(variables = ['Apti. score'], 
                         evidence = {'IQ level':0})
print(prob)
```

```python
prob = model_infer.query(variables = ['Apti. score'], 
                         evidence = {'IQ level':1})
print(prob)
```

```python
prob = model_infer.query(variables = ['IQ level'])
print(prob)
```

```python
prob = model_infer.query(variables = ['Exam level'])
print(prob)
```
